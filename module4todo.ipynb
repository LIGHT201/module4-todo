# 16.4 Use the sqlite3 module to create a SQLite database called books.db and a table called books with these fields: title (text), author (text), and year (integer).

```python
import sqlite3 as sq

# Connect to database file (creates it if it doesn't exist)
conn = sq.connect('books.db')
cursor = conn.cursor()

# Create table
cursor.execute('''
CREATE TABLE IF NOT EXISTS books (
    title TEXT,
    author TEXT,
    year INTEGER
)
''')

# Commit changes and close connection
conn.commit()
conn.close()
```

# 16.5 Read books2.csv and insert its data into the book table.

```python
import csv
import sqlite3 as sq

# Connect to your existing database
conn = sq.connect('books.db')
cursor = conn.cursor()

# Open the CSV file and read its contents
with open('books2.csv', 'r', encoding='utf-8', newline='') as file:
    reader = csv.reader(file)
    
    next(reader, None)
    
    for row in reader:
        cursor.execute('''
            INSERT INTO books (title, author, year)
            VALUES (?, ?, ?)
        ''', row)

# Save (commit) the changes and close the connection
conn.commit()
conn.close()
```

# 16.8 Use the sqlalchemy module to connect to the sqlite3 database books.db that you just made in exercise 16.4. As in 16.6, select and print the title column from the book table in alphabetical order.

```python
import sqlalchemy as sa

engine = sa.create_engine('sqlite:///books.db')

with engine.connect() as conn:
    result = conn.execute(sa.text('SELECT title FROM books ORDER BY title'))
    
    for row in result:
        print(row[0])
```
